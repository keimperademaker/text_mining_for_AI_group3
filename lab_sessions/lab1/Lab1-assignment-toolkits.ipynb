### Difference between Constituency Parsing and Dependency Parsing

**Constituency Parsing (NLTK):**
- Divides the sentence into phrases (constituents) hierarchically
- Generates a tree where phrases are grouped at different levels
- Example: "(NP the cat) (VP (V saw) (NP the dog))"
- Focus: **phrase structure**

**Dependency Parsing (spaCy):**
- Identifies relationships between individual words
- Each word has a "head" (dependent word)
- Example: "cat" depends on "saw" (nsubj), "dog" depends on "saw" (dobj)
- Focus: **relationships between words**


```python
# Select an interesting sentence for analysis
# We will use the second sentence (index 1)
sentence_idx = 1
selected_sentence_nltk = sentences_nltk[sentence_idx]
selected_tokens = tokens_per_sentence[sentence_idx]

print("SELECTED SENTENCE:")
print(selected_sentence_nltk)
print("\nTOKENS:")
print(selected_tokens)
```

```python
# Descargar el tagger de NLTK si no está disponible
nltk.download('averaged_perceptron_tagger_eng')
```

```python
# CONSTITUENCY PARSING WITH NLTK
print("\n" + "="*60)
print("CONSTITUENCY PARSING WITH NLTK")
print("="*60)

# Perform POS tagging for the tokens
pos_tags = nltk.pos_tag(selected_tokens)
print("\nPOS Tags:")
print(pos_tags)

# Use the constituency parser
constituent_parser = nltk.RegexpParser('''
NP: {<DT>? <JJ>* <NN.*>+} # Noun Phrase
VP: {<V.*> <NP>*}          # Verb Phrase
PP: {<IN> <NP>}            # Prepositional Phrase
''')

# Perform constituency parsing
tree = constituent_parser.parse(pos_tags)
print("\nCONSTITUENCY TREE (NLTK):")
print(tree)
print("\nVisualizing the tree:")
tree.pretty_print()
```

```python
# DEPENDENCY PARSING WITH SPACY
print("\n" + "="*60)
print("DEPENDENCY PARSING WITH SPACY")
print("="*60)

# Get the corresponding sentence from spaCy
selected_sentence_spacy = sents_spacy[sentence_idx]
print(f"\nSentence: {selected_sentence_spacy}")

# Show dependencies for each token
print("\nDependencies (token -> head):")
for token in selected_sentence_spacy:
    print(f"{token.text:15} -> {token.head.text:15} ({token.dep_})")

# Visualize the dependency tree
print("\nVisualizing the dependency tree:")
from spacy import displacy
displacy.render(selected_sentence_spacy, style='dep', jupyter=True)
```

### NLTK vs spaCy Comparison

**KEY DIFFERENCES:**

1. **Output Structure:**
   - NLTK: Hierarchical tree of constituents (Tree object)
   - spaCy: Linear relationships between tokens (dependency arcs)

2. **Information Captured:**
   - NLTK: Groups words into phrases (NP, VP, PP, etc.)
   - spaCy: Shows grammatical relationships (nsubj, dobj, amod, etc.)

3. **Visualization:**
   - NLTK: Tree with hierarchical structure
   - spaCy: Directed graph with dependencies

4. **Accuracy:**
   - NLTK: Depends on the RegExp Grammar you define
   - spaCy: Uses trained ML models (more accurate but less controllable)

5. **Use Cases:**
   - NLTK: Better for structural phrase analysis
   - spaCy: Better for extracting grammatical relationships and dependencies
